# Examples of API access of SurfDrive data

```python
import json
import os
import requests
from dotenv import load_dotenv
from pathlib import Path
from requests.auth import HTTPBasicAuth
from webdav4.client import Client

load_dotenv()
```

```python
USERNAME = os.getenv("SURFDRIVE_USERNAME") # SurfDrive login email address, stored in file .env
APP_PASSWD = os.getenv("SURFDRIVE_APP_PASSWD") # create on SurfDrive at User / Settings / Security
BASE_URL = f"https://surfdrive.surf.nl/remote.php/dav/files/{USERNAME}/"
```

## 1. Fetch JSON file, store it in list of dictionaries

```python
FILENAME = "deeds.json"
DIRECTORY = "rags2riches/bhic/Breda/deel_1/"

response = requests.get(BASE_URL + DIRECTORY + FILENAME, auth=HTTPBasicAuth(USERNAME, APP_PASSWD))
response.raise_for_status()
list_of_dicts = json.loads(response.text)

list_of_dicts[0]
```

## 2 Fetch JPG file, store it on disk

```python
FILENAME = "MFF-Breda-1925-1927-18_00001.jpg"
DIRECTORY = f"rags2riches/bhic/Breda/deel_1925-1927/"

response = requests.get(BASE_URL + DIRECTORY + FILENAME, auth=HTTPBasicAuth(USERNAME, APP_PASSWD))
response.raise_for_status()
with open(FILENAME, "wb") as f:
    f.write(response.content)

assert Path(FILENAME).exists()
```

## 3 Fetch list of files in directory

```python
DIRECTORY = f"rags2riches/bhic"

client = Client(BASE_URL, auth=(USERNAME, APP_PASSWD))
file_list = client.ls(DIRECTORY)

file_list[0]["name"]
```
